# Phase5

```python
import pandas as pd
df = pd.read_excel("Online_Retail_Practice.xlsx")
```

## 1. Calculate missing-value count and percentage

```python
missing_count = df.isnull().sum()
missing_percentage = (df.isnull().mean() * 100).round(2)

print("Missing-value count:")
print(missing_count)

print("\nMissing-value percentage:")
print(missing_percentage)
```

## 2. Rank columns by missing-value percentage

```python
missing_rank = (
    df.isnull().mean().mul(100)
    .sort_values(ascending=False)
    .round(2)
)

print(missing_rank)
```

## 3. Display records with missing CustomerID

```python
missing_customers = df[df["CustomerID"].isnull()]

print(missing_customers)
print("Affected transactions:", len(missing_customers))
print("Unique invoices affected:",
      missing_customers["InvoiceNo"].nunique())
```

## 4. Find missing or whitespace-only descriptions

```python
desc_col = (
    "ProductDescription"
    if "ProductDescription" in df.columns
    else "Description"
)

description = df[desc_col].astype("string")

missing_desc = df[
    description.isna() | description.str.strip().eq("").fillna(False)
]

print(missing_desc)
print("Affected rows:", len(missing_desc))
```

## 5. Create a missing-value report

```python
missing_report = pd.DataFrame({
    "Column Name": df.columns,
    "Missing Count": df.isnull().sum().values,
    "Non-Missing Count": df.notnull().sum().values,
    "Missing Percentage": (
        df.isnull().mean().mul(100).round(2).values
    )
})

print(missing_report)
```

```python
missing_report.to_excel(
    "missing_value_report.xlsx", index=False
)
```

## 6. Identify exact duplicate rows

```python
duplicate_count = df.duplicated().sum()

print("Exact duplicate rows:", duplicate_count)
```

## 7. Remove exact duplicates from a working copy

```python
df_clean = df.copy()

rows_before = len(df_clean)

df_clean = df_clean.drop_duplicates()

rows_after = len(df_clean)

print("Rows before:", rows_before)
print("Rows after:", rows_after)
print("Rows removed:", rows_before - rows_after)
print("Remaining duplicates:", df_clean.duplicated().sum())
```

## 8. Investigate repeated InvoiceNo and StockCode combinations

```python
repeated_items = df[
    df.duplicated(
        subset=["InvoiceNo", "StockCode"],
        keep=False
    )
].sort_values(["InvoiceNo", "StockCode"])

print(repeated_items)
print("Rows in repeated combinations:", len(repeated_items))
```

## 9. Handle missing descriptions and customer IDs

```python
# Convert whitespace-only descriptions to missing values
df_clean[desc_col] = (
    df_clean[desc_col]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)

# Fill missing descriptions with a documented label
df_clean[desc_col] = df_clean[desc_col].fillna(
    "Description unavailable"
)

print(df_clean[desc_col].isnull().sum())
print(df_clean["CustomerID"].isnull().sum())
```

## 10. Create a before-and-after data-quality report

```python
# Original dataset statistics
before_rows = len(df)
before_missing = df.isnull().sum().sum()
before_duplicates = df.duplicated().sum()

# Cleaned dataset statistics
after_rows = len(df_clean)
after_missing = df_clean.isnull().sum().sum()
after_duplicates = df_clean.duplicated().sum()

quality_report = pd.DataFrame({
    "Metric": [
        "Total rows",
        "Total missing values",
        "Exact duplicate rows"
    ],
    "Before Cleaning": [
        before_rows,
        before_missing,
        before_duplicates
    ],
    "After Cleaning": [
        after_rows,
        after_missing,
        after_duplicates
    ]
})

print(quality_report)

quality_report.to_excel(
    "data_quality_report.xlsx", index=False
)
```

```python
df_clean["Revenue"] = (
    df_clean["Quantity"] * df_clean["UnitPrice"]
)

total_revenue = df_clean["Revenue"].sum()

print("Total Revenue:", total_revenue)
```
