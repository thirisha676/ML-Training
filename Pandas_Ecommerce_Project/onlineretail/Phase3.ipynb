# Phase3

```python
import pandas as pd
df = pd.read_excel("Online_Retail_Practice.xlsx")
df
```

## 1. Display only InvoiceNo and InvoiceDate

```python

print(df[["InvoiceNo", "InvoiceDate"]])
```

# 2. Retrieve a transaction using iloc

```python

print(df.iloc[5])
```

## 3. Use loc to select rows and columns

```python

print(df.loc[0:5, ["InvoiceNo", "Country"]])
```

## 4. Filter transactions where Quantity is greater than 10

```python

print(df[df["Quantity"] > 10])
```

## 5. Find transactions where UnitPrice is greater than £50

```python

print(df[df["UnitPrice"] > 50])
```

## 6. Retrieve transactions from a specific country

```python

print(df[df["Country"] == "France"])
```

## 7. Find transactions from the five busiest countries using isin()

```python

top5 = df["Country"].value_counts().head(5).index

print(df[df["Country"].isin(top5)])
```

## 8. Find quantities between 5 and 20

```python

print(df[df["Quantity"].between(5, 20)])
```

## 9. Filter using query()

```python

print(df.query("Quantity > 10 and UnitPrice > 5"))
```

## 10. Sort by UnitPrice and display the 20 highest prices

```python

print(df.sort_values("UnitPrice", ascending=False).head(20))
```
