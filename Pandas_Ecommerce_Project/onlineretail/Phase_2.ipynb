## 6. Count distinct invoices, products, countries, and customers

```python

print("Unique invoices:", df["InvoiceNo"].nunique())
print("Unique products:", df["StockCode"].nunique())
print("Unique countries:", df["Country"].nunique())
print("Unique customers:", df["CustomerID"].nunique())
```

## 7. Display unique countries and count them

```python

print(df["Country"].unique())

print("Number of countries:", df["Country"].nunique())
```

## 8. Display the 10 countries with the most transactions

```python

print(df["Country"].value_counts().head(10))
```

## 9. Create a compact dataset summary

```python

summary = pd.DataFrame({
    "Total Rows": [df.shape[0]],
    "Total Columns": [df.shape[1]],
    "Unique Products": [df["StockCode"].nunique()],
    "Unique Invoices": [df["InvoiceNo"].nunique()]
})

print(summary)
```

## 10. Export the dataset profile to CSV

```python

summary.to_csv("dataset_profile.csv", index=False)

print("CSV file saved successfully!")
```
