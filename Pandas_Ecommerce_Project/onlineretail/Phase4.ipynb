# 1. Create a working copy and keep the original safe

```python

import pandas as pd
df = pd.read_excel("Online_Retail_Practice.xlsx")
```

```python

df_raw = df.copy()
df = df_raw.copy()

print(df.head())
```

```python

df = df.rename(columns={"Description": "ProductDescription"})

print(df.head())
```

## 2. Rename Description to ProductDescription

```python

df = df.rename(columns={"Description": "ProductDescription"})

print(df.head())
```

## 3. Create a Revenue column

```python

df["Revenue"] = df["Quantity"] * df["UnitPrice"]

print(df[["Quantity", "UnitPrice", "Revenue"]].head())
```

## 4. Create a TransactionType column

```python

import numpy as np

conditions = [
    df["InvoiceNo"].astype("string").str.startswith("C", na=False),
    df["Quantity"] < 0,
    df["Quantity"] > 0
]

choices = ["Cancellation", "Return", "Positive sale"]

df["TransactionType"] = np.select(
    conditions,
    choices,
    default="Other/invalid"
)

print(df[["InvoiceNo", "Quantity", "TransactionType"]].head(10))
```

## 5. Create IsIdentifiedCustomer

```python

df["IsIdentifiedCustomer"] = df["CustomerID"].notna()

print(df[["CustomerID", "IsIdentifiedCustomer"]].head())
```

## 6. Convert InvoiceDate to datetime

```python

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

print(df["InvoiceDate"].dtype)
```

## 7. Convert columns to string data types

```python

df["InvoiceNo"] = df["InvoiceNo"].astype("string")
df["StockCode"] = df["StockCode"].astype("string")
df["Country"] = df["Country"].astype("string")

print(df[["InvoiceNo", "StockCode", "Country"]].dtypes)
```

## 8. Reorder the columns

```python

columns = [
    "InvoiceNo", "InvoiceDate", "StockCode",
    "ProductDescription", "CustomerID", "Country",
    "Quantity", "UnitPrice", "Revenue",
    "TransactionType", "IsIdentifiedCustomer"
]

df = df[columns]

print(df.head())
```

## 9. Sort by InvoiceDate and InvoiceNo

```python

df = df.sort_values(
    by=["InvoiceDate", "InvoiceNo"]
)

print(df.head())
```

## 10. Reset and verify the index

```python

df = df.reset_index(drop=True)

print(df.head())
print("Sequential index:", df.index.equals(pd.RangeIndex(len(df))))
```

```python
df.to_excel("cleaned_data.xlsx", index=False)

print("Excel file saved successfully!")
```
